```python
def show_grid(folder, n=6, figsize=(12,8)):
    paths = glob.glob(os.path.join(folder, "*.jpg"))
    sample = random.sample(paths, min(n, len(paths)))
    plt.figure(figsize=figsize)
    for i,p in enumerate(sample,1):
        im = cv2.cvtColor(cv2.imread(p), cv2.COLOR_BGR2RGB)
        plt.subplot(1, n, i); plt.imshow(im); plt.title(os.path.basename(p)); plt.axis('off')
    plt.show()

show_grid(".\Yolo\data\Val\Glioma\images", n=5)
```
